```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use(['science', 'notebook', 'grid'])
```

```python
u = np.load('burgers_data.npz', allow_pickle=True)
```

```python
xx, tt = np.meshgrid(u['x'], u['t'], indexing='ij')
```

```python
xx.shape, tt.shape, u['u'].shape
```

```python
xx, tt
```

```python
xx[:, 0], tt[:, -1]
```

```python
plt.plot(u['x'], u['u'][:, -1]);
```

```python
plt.contourf(xx, tt, u['u'], 1000, cmap='rainbow')
plt.xlabel(r'$x\, [\, - \,]$')
plt.ylabel(r'$t\, [\, - \,]$')
plt.title(r'$u(\, x, t\,)_{rk4}$')
plt.colorbar()
plt.savefig('burgers_entire.pdf')
```
